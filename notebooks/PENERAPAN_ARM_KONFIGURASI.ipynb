```python
# LOAD DATA
FILE = "DATA_GREEN_FLOWER_PERFECTED.xlsx"
if not os.path.exists(FILE):
    from google.colab import files
    up = files.upload()
    FILE = list(up)[0]

df = pd.read_excel(FILE, sheet_name="Data_Perfected")
print("Shape:", df.shape)
print("Kolom:", df.columns.tolist())
df["Nama_Barang"] = df["Nama_Barang"].astype(str).str.strip()
df["Tanggal"] = pd.to_datetime(df["Tanggal"], format="%Y-%m-%d")
df["Bulan"] = df["Tanggal"].dt.to_period("M").astype(str)
```

```python
# PEMERIKSAAN DATA
print("Shape:", df.shape)
print("Kolom:", df.columns.tolist())
print("Missing value total:", int(df.isna().sum().sum()))
print("Baris duplikat exact:", int(df.duplicated().sum()),
      "| pasangan ID+Barang berulang:", int(df.duplicated(["ID", "Nama_Barang"]).sum()))
print("Jumlah transaksi (ID unik):", df["ID"].nunique())
print("Jumlah item unik:", df["Nama_Barang"].nunique())
print("Periode:", df["Tanggal"].min().date(), "s.d.", df["Tanggal"].max().date())
print("Setiap ID berada di satu bulan:", bool((df.groupby("ID")["Bulan"].nunique() == 1).all()))
print("\nFrekuensi baris per item:\n", df["Nama_Barang"].value_counts())
```

```python
#  TRANSACTION BASKET
# 1 transaksi = kumpulan item unik dalam satu ID
trx = df.groupby("ID").agg(items=("Nama_Barang", lambda x: sorted(set(x))),
                           Bulan=("Bulan", "first"))
te = TransactionEncoder()
basket = pd.DataFrame(te.fit(trx["items"]).transform(trx["items"]),
                      columns=te.columns_, index=trx.index)
N = len(basket)
print("\nJumlah transaksi:", N, "| jumlah item:", basket.shape[1])
print("Transaksi per bulan:\n", trx["Bulan"].value_counts().sort_index())
print("Support per item:\n", basket.mean().sort_values(ascending=False).round(4))
```

```python
# FUNGSI APRIORI
def run(data, min_sup, min_conf):
    fi = apriori(data, min_support=min_sup, use_colnames=True)
    if fi.empty:
        return 0, pd.DataFrame()
    rules = association_rules(fi, metric="confidence", min_threshold=min_conf)
    if rules.empty:
        return len(fi), pd.DataFrame()
    rules["rule"] = rules.apply(lambda r: " + ".join(sorted(r["antecedents"])) + " → "
                                + " + ".join(sorted(r["consequents"])), axis=1)
    rules["n_trx"] = (rules["support"] * len(data)).round().astype(int)
    rules = rules[["rule", "n_trx", "support", "confidence", "lift"]]
    return len(fi), rules.sort_values("lift", ascending=False).reset_index(drop=True)
```

```python
# GLOBAL ARM: DUA KONFIGURASI
fi1, rules1 = run(basket, 0.05, 0.50)
fi2, rules2 = run(basket, 0.02, 0.30)

print("\n=== GLOBAL 1 (support 0.05, confidence 0.50) ===")
display(rules1.round(4))
print("\n=== GLOBAL 2 (support 0.02, confidence 0.30) ===")
display(rules2.round(4))
```

```python
# MONTHLY ARM
bulan_list = sorted(trx["Bulan"].unique())
rows, ringkas = [], []

for b in bulan_list:
    data_b = basket[trx["Bulan"] == b]
    _, r = run(data_b, 0.02, 0.30)

    ringkas.append({
        "Bulan": b,
        "Transaksi": len(data_b),
        "Jumlah Rules": len(r)
    })

    if len(r):
        rows.append(r.assign(Bulan=b))

monthly = pd.concat(rows, ignore_index=True)

print("\n=== RINGKASAN MONTHLY ARM ===")
display(pd.DataFrame(ringkas))

print("\n=== RULE BULANAN (RQ2) ===")
display(
    monthly[["Bulan", "rule", "support", "confidence", "lift"]]
    .sort_values(["Bulan", "confidence"], ascending=[True, False])
    .round(4)
)
```

```python
# RULE STABILITY
# Stability = jumlah bulan rule muncul / jumlah seluruh bulan pengamatan
TOTAL_BULAN = len(bulan_list)
stab = (monthly.groupby("rule")
        .agg(Bulan_Muncul=("Bulan", "nunique"),
             Rata_Support=("support", "mean"),
             Rata_Confidence=("confidence", "mean"),
             Rata_Lift=("lift", "mean"),
             Bulan_Lift_gt1=("lift", lambda s: int((s > 1).sum())),
             Bulan_min2trx=("n_trx", lambda s: int((s >= 2).sum())))
        .reset_index())
stab["Total_Bulan"] = TOTAL_BULAN
stab["Stability_%"] = (stab["Bulan_Muncul"] / TOTAL_BULAN * 100).round(1)

# Stability untuk rule hasil Global 2 (mencakup seluruh rule Global 1)
tabel_stab = (rules2[["rule", "n_trx", "support", "confidence", "lift"]]
              .rename(columns={"support": "Support_Global", "confidence": "Confidence_Global", "lift": "Lift_Global"})
              .merge(stab, on="rule", how="left"))
for c in ["Bulan_Muncul", "Bulan_Lift_gt1", "Bulan_min2trx"]:
    tabel_stab[c] = tabel_stab[c].fillna(0).astype(int)
tabel_stab["Total_Bulan"] = TOTAL_BULAN
tabel_stab["Stability_%"] = (tabel_stab["Bulan_Muncul"] / TOTAL_BULAN * 100).round(1)
tabel_stab = tabel_stab.sort_values(["Bulan_Muncul", "Lift_Global"], ascending=False).reset_index(drop=True)
print("\n=== RULE STABILITY (rule Global 2) ===")
print("Bulan_min2trx = jumlah bulan rule didukung >= 2 transaksi (pengaman: pada support 0.02 satu transaksi sudah cukup)")
display(tabel_stab.round(4))
```

```python
# PEMILIHAN ATURAN
kandidat = tabel_stab[
    (tabel_stab["Lift_Global"] > 1) &
    (tabel_stab["Bulan_Muncul"] >= 6)
].copy()

kandidat = kandidat.sort_values(
    ["Bulan_Muncul", "Lift_Global"],
    ascending=False
).reset_index(drop=True)

# Ambil 10 aturan kandidat
kandidat_final = kandidat.head(10)

print("\n=== ATURAN TERPILIH ===")

display(
    kandidat_final[
        [
            "rule",
            "Support_Global",
            "Confidence_Global",
            "Lift_Global",
            "Stability_%"
        ]
    ].round(4)
)
```
